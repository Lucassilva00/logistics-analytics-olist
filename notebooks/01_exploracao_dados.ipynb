## 13. Análise regional das entregas

```python
analise_entrega_regiao = analise_entrega_satisfacao.merge(
    customers[["customer_id", "customer_state"]],
    on="customer_id",
    how="left",
    validate="many_to_one"
)
```

```python
print("Linhas:", len(analise_entrega_regiao))

print(
    "Pedidos únicos:",
    analise_entrega_regiao["order_id"].nunique()
)

print(
    "Pedidos sem estado:",
    analise_entrega_regiao["customer_state"].isna().sum()
)
```

```python
resumo_regional = analise_entrega_regiao.groupby("customer_state").agg(
    pedidos_classificados=("order_id", "count"),
    pedidos_atrasados=(
        "status_entrega",
        lambda x: (x == "Atrasado").sum()
    ),
    pedidos_no_prazo=(
        "status_entrega",
        lambda x: (x == "No prazo").sum()
    )
)
```

```python
resumo_regional["taxa_atraso_pct"] = (
    resumo_regional["pedidos_atrasados"]
    / resumo_regional["pedidos_classificados"]
    * 100
)

resumo_regional["taxa_no_prazo_pct"] = (
    resumo_regional["pedidos_no_prazo"]
    / resumo_regional["pedidos_classificados"]
    * 100
)
```

```python
resumo_regional.sort_values(
    "pedidos_classificados",
    ascending=False
).head(10).round(2)
```

```python
estados_volume_relevante = resumo_regional[
    resumo_regional["pedidos_classificados"] >= 500
]

estados_maior_atraso = estados_volume_relevante.sort_values(
    "taxa_atraso_pct",
    ascending=False
)

estados_maior_atraso.head(10).round(2)
```

```python
pedidos_atrasados_regiao = analise_entrega_regiao[
    analise_entrega_regiao["status_entrega"] == "Atrasado"
]
```

```python
len(pedidos_atrasados_regiao)
```

```python
resumo_severidade = pedidos_atrasados_regiao.groupby(
    "customer_state"
).agg(
    atraso_medio_dias=("dias_atraso", "mean"),
    atraso_mediano_dias=("dias_atraso", "median"),
    maior_atraso_dias=("dias_atraso", "max")
)
```

```python
resumo_regional_detalhado = resumo_regional.join(
    resumo_severidade
)

resumo_regional_detalhado[
    resumo_regional_detalhado["pedidos_classificados"] >= 500
].sort_values(
    "taxa_atraso_pct",
    ascending=False
).head(10).round(2)
```

## 15. Definição do modelo analítico

### Objetivo

Organizar os dados da Olist em um modelo analítico no PostgreSQL, preparado para responder às perguntas de negócio do projeto Logistics Analytics e alimentar o dashboard no Power BI.

O modelo deverá preservar os dados originais, registrar as regras de tratamento e permitir análises de prazo de entrega, satisfação dos clientes, distribuição regional dos pedidos e desempenho comercial das categorias de produtos.

### Estrutura geral

O modelo será composto por duas tabelas de fatos e três dimensões principais.

**Tabelas de fatos**

- `fact_order_delivery`: uma linha por `order_id`, com informações do ciclo do pedido, datas, indicadores de prazo e avaliações consolidadas.
- `fact_order_item`: uma linha por combinação de `order_id` e `order_item_id`, com informações de produto, vendedor, preço e frete.

**Dimensões**

- `dim_customer`: dados de identificação e localização dos clientes.
- `dim_product`: características dos produtos, categoria original e tradução disponível.
- `dim_seller`: dados de identificação e localização dos vendedores.

### Grão das tabelas

O grão define o que cada linha representa em uma tabela.

- `fact_order_delivery`: um registro por pedido. A fonte contém 99.441 pedidos.
- `fact_order_item`: um registro por item do pedido. A fonte contém 112.650 itens.
- `dim_customer`: um registro por `customer_id`. A fonte contém 99.441 registros.
- `dim_product`: um registro por `product_id`. A fonte contém 32.951 produtos.
- `dim_seller`: um registro por `seller_id`. A fonte contém 3.095 vendedores.

### Princípios de modelagem

As tabelas de fatos manterão seus respectivos grãos. Os relacionamentos serão definidos por chaves primárias e estrangeiras, evitando multiplicações indevidas de registros.

As transformações serão feitas sobre os dados carregados em memória ou em etapas específicas de preparação, sem modificar os arquivos originais de `data/raw`.

Os campos derivados, como atraso em dias, classificação da entrega e nota média por pedido, serão documentados com suas regras de cálculo e tratamento de valores ausentes.

### 15.1 Regras de relacionamento e transformação

#### 1. Associação entre pedidos e itens

O arquivo `olist_order_items_dataset.csv` não possui a coluna `customer_id`. Para incluir essa informação em `fact_order_item`, será necessário relacionar `order_items` com `orders` usando `order_id`.

#### 2. Relacionamento entre as tabelas de fatos

`fact_order_delivery` terá uma linha por pedido (`order_id`), enquanto `fact_order_item` terá uma linha por item (`order_id` + `order_item_id`).

As tabelas de fatos não serão relacionadas diretamente no Power BI. As dimensões serão utilizadas para organizar os filtros e evitar duplicações ou ambiguidades nas análises.

#### 3. Tratamento dos CEPs

Os prefixos de CEP serão armazenados como texto de cinco caracteres, preservando zeros à esquerda. Essa padronização será realizada durante a preparação dos dados, sem alterar os arquivos originais.

#### 4. Tratamento das categorias

A tradução disponível em `product_category_name_translation` será incorporada à dimensão de produtos.

Categorias sem tradução manterão sua identificação original. Produtos sem categoria preenchida serão identificados como `Sem categoria` na camada analítica, sem modificar os dados brutos.

#### 5. Indicadores de prazo de entrega

`dias_atraso` será calculado pela diferença entre a data real de entrega ao cliente e a data estimada, considerando somente a data, sem o horário.

A classificação será:
- `No prazo`: entrega realizada até a data estimada.
- `Atrasado`: entrega realizada depois da data estimada.
- Valor ausente: quando não houver informação suficiente para classificar o prazo.

Pedidos não entregues e pedidos sem data real de entrega não serão classificados artificialmente como entregues no prazo.

#### 6. Tempo entre transportadora e cliente

`tempo_transportadora_cliente` será calculado em dias inteiros somente quando as duas datas estiverem disponíveis e a data de entrega ao cliente não for anterior à data registrada de entrega à transportadora.

Registros sem as datas necessárias ou com sequência temporal inconsistente permanecerão sem valor para essa métrica específica. Os dados originais serão preservados.

#### 7. Consolidação das avaliações

As avaliações de `order_reviews` serão consolidadas por `order_id`, produzindo uma nota média e a quantidade de avaliações registradas para cada pedido.

Pedidos sem avaliação terão nota média ausente e quantidade de avaliações igual a zero. A média consolidada será identificada como uma métrica calculada, e não como uma nota original da fonte.

#### 8. Preservação dos dados originais

Os arquivos de `data/raw` permanecerão inalterados. As transformações serão realizadas em DataFrames ou em etapas específicas de preparação e carga no PostgreSQL, com regras documentadas e validações de qualidade.

```python
from sqlalchemy import text
import pandas as pd

consulta_satisfacao_sql = text("""
    WITH pedidos_com_satisfacao AS (
        SELECT
            status_entrega,

            CASE
                WHEN nota_media_pedido <= 2 THEN 'Baixa'
                WHEN nota_media_pedido < 4 THEN 'Intermediária'
                ELSE 'Alta'
            END AS faixa_satisfacao

        FROM public.fact_order_delivery

        WHERE status_entrega IS NOT NULL
          AND nota_media_pedido IS NOT NULL
    ),

    resumo AS (
        SELECT
            status_entrega,
            faixa_satisfacao,
            COUNT(*) AS quantidade_pedidos

        FROM pedidos_com_satisfacao

        GROUP BY
            status_entrega,
            faixa_satisfacao
    )

    SELECT
        status_entrega,
        faixa_satisfacao,
        quantidade_pedidos,

        ROUND(
            100.0 * quantidade_pedidos
            / SUM(quantidade_pedidos) OVER (
                PARTITION BY status_entrega
            ),
            2
        ) AS percentual_pct

    FROM resumo

    ORDER BY
        CASE status_entrega
            WHEN 'Atrasado' THEN 1
            WHEN 'No prazo' THEN 2
            ELSE 3
        END,
        CASE faixa_satisfacao
            WHEN 'Baixa' THEN 1
            WHEN 'Intermediária' THEN 2
            WHEN 'Alta' THEN 3
            ELSE 4
        END;
""")

with engine.connect() as conexao:
    resumo_satisfacao_sql = pd.read_sql(
        consulta_satisfacao_sql,
        conexao
    )

display(resumo_satisfacao_sql)
```

### 18.7 Auditoria complementar da qualidade dos dados

Durante a validação do dashboard, foram identificados dois comportamentos que exigiram uma investigação adicional na fonte original: a redução abrupta do volume de pedidos nos meses finais de 2018 e a repetição de identificadores na tabela de avaliações (`order_reviews`).

O objetivo desta auditoria foi verificar a origem desses comportamentos, avaliar possíveis inconsistências nas associações entre pedidos e avaliações e medir o impacto das ocorrências sobre os indicadores de satisfação.

As verificações foram realizadas sobre cópias e consultas diagnósticas, preservando os dados originais.

#### 18.7.1 Investigação da distribuição temporal dos pedidos

A análise mensal identificou uma redução acentuada no final da série:

- Julho/2018: 6.292 pedidos.
- Agosto/2018: 6.512 pedidos.
- Setembro/2018: 16 pedidos.
- Outubro/2018: 4 pedidos.

A última compra registrada ocorreu em 17/10/2018.

#### 18.7.2 Auditoria dos identificadores de avaliação

A tabela `order_reviews` possui 99.224 linhas e 98.410 identificadores distintos de avaliação.

Foram encontrados 789 valores de `review_id` associados a mais de um pedido, envolvendo 1.603 linhas e 1.412 pedidos distintos.

A ocorrência de um mesmo identificador em pedidos diferentes indica que `review_id` não deve ser utilizado isoladamente como chave única. Por isso, a consolidação das avaliações foi realizada por `order_id`.

#### 18.7.3 Consistência temporal das avaliações

A comparação entre a data de criação da avaliação e a data de compra identificou 64 associações em que a avaliação antecedia a compra:

- 57 relacionadas a pedidos cancelados;
- 6 relacionadas a pedidos entregues;
- 1 relacionada a pedido enviado.

A comparação foi realizada considerando somente a data do calendário, sem considerar o horário, pois `review_creation_date` não possui precisão horária.

#### 18.7.4 Investigação das associações entre avaliações e pedidos

Foi testada uma regra diagnóstica que considerava como candidatos os pedidos entregues cuja data de compra fosse igual ou anterior à data da avaliação.

Entre os 789 identificadores repetidos:

- 84 possuíam exatamente um candidato;
- 11 não possuíam nenhum candidato;
- 694 possuíam mais de um candidato.

Os resultados demonstraram que não há evidência suficiente para reconstruir automaticamente as associações originais. Nenhuma reassociação foi realizada.

#### 18.7.5 Testes de sensibilidade dos indicadores de satisfação

Foram realizados dois testes diagnósticos.

No primeiro, foram temporariamente retiradas as 64 linhas com avaliações anteriores à compra. A nota média passou de 4,2906 para 4,2907 nos pedidos no prazo e permaneceu em 2,2718 nos pedidos atrasados.

Esses testes foram utilizados para avaliar a sensibilidade dos indicadores, não para definir uma exclusão definitiva de registros.

#### 18.7.6 Decisões metodológicas e limitações

Com base nas verificações, foram adotadas as seguintes decisões:

1. Preservar os arquivos originais e todos os registros de pedidos.

3. Não realizar reassociação automática de avaliações a pedidos.
